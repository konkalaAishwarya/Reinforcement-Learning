```python

n = int(input("Enter number of machines: "))
machines = {}
for i in range(n):
    name = input(f"Enter name of machine {i + 1}: ")
    rewards = list(map(float, input(
        f"Enter observed rewards for {name} (space separated): "
    ).split()))

    machines[name] = rewards
q_values = {}

for machine, rewards in machines.items():
    if len(rewards) > 0:
        q_values[machine] = sum(rewards) / len(rewards)
    else:
        q_values[machine] = 0
print("\nEstimated Values Q(a):")

for machine, q in q_values.items():
    print(f"{machine}: {q:.2f}")
selected_machine = max(q_values, key=q_values.get)

print("\nGreedy Algorithm Selection:")
print(f"Selected Machine: {selected_machine}")
print(f"Q({selected_machine}) = {q_values[selected_machine]:.2f}")
```

```python
n = int(input("Enter number of machines: "))

q_values = {}
counts = {}
machines = {}

for i in range(n):
    name = input(f"Enter machine {i + 1} name: ")
    rewards = list(map(float, input(
        f"Enter observed rewards for {name}: "
    ).split()))

    machines[name] = rewards
    q_values[name] = 0
    counts[name] = 0

print("\nIncremental Updates:")

for machine, rewards in machines.items():
    for reward in rewards:
        counts[machine] += 1

        q_values[machine] = (
            q_values[machine]
            + (reward - q_values[machine]) / counts[machine]
        )

        print(
            f"Machine: {machine}, "
            f"Reward: {reward:g}, "
            f"N: {counts[machine]}, "
            f"Q: {q_values[machine]:.2f}"
        )

        selected_machine = max(q_values, key=q_values.get)

        print(
            f"Greedy Selection: {selected_machine} "
            f"(Q = {q_values[selected_machine]:.2f})"
        )

print("\nFinal Q Values:")

for machine in q_values:
    print(f"{machine}: {q_values[machine]:.2f}")

selected_machine = max(q_values, key=q_values.get)

print("\nFinal Greedy Selection:")
print(f"Selected Machine: {selected_machine}")
print(f"Q({selected_machine}) = {q_values[selected_machine]:.2f}")
```
