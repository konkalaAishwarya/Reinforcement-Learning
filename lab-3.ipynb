```python
import random

# ============================================================
# Modified Epsilon-Greedy Multi-Armed Bandit
# ============================================================

# Part A - Machine Initialization
machines = ['A', 'B', 'C']

M = 3
N_play = 10
epsilon = 0.2

# Part B - Initialize knowledge tables
N = [0, 0, 0]       # Number of times each machine is selected
R = [0, 0, 0]       # Total reward obtained from each machine
Q = [0.0, 0.0, 0.0] # Estimated reward of each machine

# Reward patterns given in the question
reward_pattern = {
    'A': [3, 4, 2, 5],
    'B': [6, 7, 5, 8],
    'C': [4, 3, 5, 4]
}

# Keep track of which reward should be used next
reward_index = {
    'A': 0,
    'B': 0,
    'C': 0
}

# Fixed seed for reproducible results
random.seed(10)

print("Modified Epsilon-Greedy Strategy")
print("---------------------------------")
print("Machines:", machines)
print("Number of plays:", N_play)
print("Epsilon:", epsilon)
print()

# ============================================================
# Part C, D and E - Decision, Reward Collection and Q Update
# ============================================================

for play in range(1, N_play + 1):

    # Generate random number
    r = random.random()

    # Find highest Q-value machine
    max_Q = max(Q)
    highest_machine = Q.index(max_Q)

    if r < epsilon:
        # Exploration:
        # Select randomly from machines EXCEPT highest Q machine
        available = [i for i in range(M) if i != highest_machine]
        selected = random.choice(available)

        decision = "Exploration"

    else:
        # Exploitation:
        # Select machine having highest Q-value
        selected = highest_machine

        decision = "Exploitation"

    machine = machines[selected]

    # Get reward from given reward pattern
    index = reward_index[machine]

    if index < len(reward_pattern[machine]):
        reward = reward_pattern[machine][index]
    else:
        # If pattern is exhausted, ask user
        while True:
            try:
                reward_input = input(f"Play with machine {machine}\nEnter reward (single number expected): ")
                reward = float(reward_input)
                break # Exit loop if conversion is successful
            except ValueError:
                print("Invalid input. Please enter a single numerical value for the reward.")

    reward_index[machine] += 1

    # Update N
    N[selected] += 1

    # Update total reward R
    R[selected] += reward

    # Incremental sample-average update
    Q[selected] = Q[selected] + (reward - Q[selected]) / N[selected]

    # Display play information
    print(f"Play {play}")
    print(f"Random number = {r:.4f}")
    print(f"Decision = {decision}")
    print(f"Selected machine = {machine}")
    print(f"Reward = {reward}")
    print(f"N = {N}")
    print(f"R = {R}")
    print(f"Q = {[round(x, 2) for x in Q]}")
    print("---------------------------------")


# ============================================================
# Part F - Final Results
# ============================================================

print("\nFINAL RESULTS")
print("=============")

print("\nParticipation Count (N):")
for i in range(M):
    print(f"Machine {machines[i]} = {N[i]}")

print("\nTotal Rewards (R):")
for i in range(M):
    print(f"Machine {machines[i]} = {R[i]}")

print("\nEstimated Q-Values:")
for i in range(M):
    print(f"Machine {machines[i]} = {Q[i]:.2f}")

# Total reward received by player
total_reward = sum(R)

print("\nTotal Reward Received =", total_reward)

# Best machine
best_machine = machines[Q.index(max(Q))]

print("Best Machine =", best_machine)
print("Highest Q-Value =", f"{max(Q):.2f}")
```
